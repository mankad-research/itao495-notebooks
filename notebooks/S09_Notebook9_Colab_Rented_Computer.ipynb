## Exercise 1 — What machine is this?

```python
import subprocess, platform, os, time

def sh(cmd):
    """Run a terminal command and print what it says."""
    out = subprocess.run(cmd, shell=True, capture_output=True, text=True)
    print((out.stdout or out.stderr).strip() or "(no output)")

print("Operating system:"); sh("cat /etc/os-release | head -2")
print("\nProcessor cores (vCPUs):"); sh("nproc")
print("\nMemory (RAM):"); sh("free -h | head -2")
print("\nDisk:"); sh("df -h / | tail -1")
print("\nProcessor model:"); sh("lscpu | grep 'Model name'")
```

## Exercise 3 — How fast is it?

A simple speed test: count the prime numbers below a limit. Run it, write down the time, and compare with the Azure VM later (you will run the same cell there if your instructor says so).

```python
# ✏️ CHANGE ME — try 1_000_000, then 10_000_000
LIMIT = 5_000_000

def count_primes(n):
    sieve = [True] * (n + 1)
    sieve[0:2] = [False, False]
    for i in range(2, int(n ** 0.5) + 1):
        if sieve[i]:
            for j in range(i * i, n + 1, i):
                sieve[j] = False
    return sum(sieve)

t0 = time.time()
count = count_primes(LIMIT)
print(f"{count:,} primes below {LIMIT:,} in {time.time() - t0:.2f} seconds")
```
