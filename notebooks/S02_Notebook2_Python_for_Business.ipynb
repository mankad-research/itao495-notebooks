## Exercise 3 — Lists and loops

A **list** holds several items in order. A **for loop** does the same thing to each item, like filling a formula down a column.

The cell below reads eight customer comments and counts the ones that mention a late delivery.

````python
comments = [
    "Package arrived late and the box was damaged.",
    "Great jacket, fits perfectly.",
    "Delivery was LATE again. Third time this year.",
    "Support was friendly and fast.",
    "Price is a bit high but quality is good.",
    "Arrived two days late, no tracking updates.",
    "The zipper broke after one week.",
    "Fast shipping, very happy!",
]

late_count = 0
for comment in comments:
    if "late" in comment.lower():
        late_count = late_count + 1
        print("LATE ->", comment)

print()
print(f"{late_count} of {len(comments)} comments mention lateness.")
````

## Exercise 4 — A table (a spreadsheet in code)

**pandas** is the standard tool for tables in Python. A table is called a **DataFrame**. Each line below does one spreadsheet-style operation: filter, add a column, summarize.

````python
import pandas as pd

sales = pd.DataFrame({
    "region":  ["East", "East", "West", "West", "South", "South", "East", "West", "South", "East", "West", "South"],
    "product": ["Widget", "Gadget", "Widget", "Gadget", "Widget", "Gadget", "Gadget", "Widget", "Widget", "Widget", "Gadget", "Gadget"],
    "units":   [120, 80, 95, 60, 70, 110, 40, 130, 55, 90, 75, 65],
    "price":   [9.5, 14.0, 9.5, 14.0, 9.5, 14.0, 14.0, 9.5, 9.5, 9.5, 14.0, 14.0],
})
print(sales.head())          # first 5 rows
print()
print("Rows and columns:", sales.shape)
````

````python
sales["revenue"] = sales["units"] * sales["price"]       # new column, like =units*price

east = sales[sales["region"] == "East"]                  # filter rows
print("East region rows:")
print(east)
print()

by_region = sales.groupby("region")["revenue"].sum()     # like a pivot table
print("Revenue by region:")
print(by_region)
print()
print("Total revenue:", sales["revenue"].sum())
````

## Exercise 8 — Find the bug

Below is code of the kind AI assistants write. **Both cells have a problem.** Find it *before* you ask anyone for help.

### Bug A: it runs, and it is wrong
The code should give the **average revenue per order** for the `tiny` table (5 orders, 1,200 total revenue). Work out the right answer by hand first.

````python
# Bug A — written by an AI assistant
average_order_value = tiny["revenue"].sum() / tiny["region"].nunique()
print("Average order value:", average_order_value)
````

<details><summary>Click to see the answer</summary>

Total revenue is 1,200 and there are 5 orders, so the average is **240**. The code divides by the number of **regions** (`nunique()` counts distinct values: 2), giving 600. It runs without any error and prints a confident, wrong number. This is the most dangerous kind of AI mistake, and the only defense is checking against a case you can work out by hand.

Fix: `tiny["revenue"].mean()` or `tiny["revenue"].sum() / len(tiny)`.
</details>

## Exercise 9 — Calling a model from your own code

Now the payoff. You have been calling functions. Calling an AI model is the same idea: you send text, you get text back. First, store your Groq key (skip if you did this in Notebook 1).

### One-time setup: store your key in Colab Secrets
1. Click the **🔑 key icon** in the left sidebar of Colab.
2. Click **Add new secret**. Name: `GROQ_API_KEY`. Value: your key (it starts with `gsk_`).
3. Turn on **Notebook access** for this notebook.

Colab will remember the key for all your notebooks, so you never paste it into a cell. If you skip this, the setup cell will simply ask you to paste the key each time (the box hides what you type).

````python
# Run this cell first. Do not modify.
!pip install groq --quiet

import time, json, textwrap
from groq import Groq
import groq

def _get_key():
    """Look for the key in Colab Secrets first; otherwise ask for it (hidden as you type)."""
    try:
        from google.colab import userdata
        return userdata.get("GROQ_API_KEY")
    except Exception:
        import getpass
        return getpass.getpass("Paste your Groq API key (starts with gsk_): ")

client = Groq(api_key=_get_key())
MODEL = "openai/gpt-oss-20b"         # if Groq retires this name, change it here only

# Groq retired llama-3.1-8b-instant for free accounts in August 2026, so this notebook uses
# openai/gpt-oss-20b. It is a reasoning model: before it answers, it writes hidden "thinking"
# tokens. They count toward max_tokens and are billed as output tokens. The lines below keep
# that thinking short and add room for it, so every max_tokens in this notebook still means
# "room for the answer".
REASONING_EFFORT = "low"      # "low", "medium" or "high"
REASONING_HEADROOM = 300      # extra tokens reserved for the hidden thinking
_groq_create = client.chat.completions.create
def _create_with_reasoning(*args, **kwargs):
    if "gpt-oss" in str(kwargs.get("model", "")):
        kwargs.setdefault("reasoning_effort", REASONING_EFFORT)
        kwargs.setdefault("include_reasoning", False)
        if kwargs.get("max_tokens"):
            kwargs["max_tokens"] += REASONING_HEADROOM
    return _groq_create(*args, **kwargs)
client.chat.completions.create = _create_with_reasoning

def _retry_after(err, default=15):
    """Seconds the free tier asks us to wait (from the error's retry-after header), capped at 60."""
    try:
        return min(60.0, float(err.response.headers.get("retry-after", default)))
    except Exception:
        return float(default)

def ask(prompt, system=None, temperature=0.0, max_tokens=300, messages=None, model=None):
    """Send one request to the model and return the full response object.
    Waits and retries if the free tier says 'slow down'."""
    if messages is None:
        messages = ([{"role": "system", "content": system}] if system else []) + [{"role": "user", "content": prompt}]
    for attempt in range(6):
        try:
            return client.chat.completions.create(model=model or MODEL, messages=messages,
                                                  temperature=temperature, max_tokens=max_tokens)
        except groq.RateLimitError as err:
            wait = _retry_after(err)
            print(f"(rate limit hit - waiting {wait:.0f} seconds, then trying again)")
            time.sleep(wait)
    raise RuntimeError("Still rate limited. Wait a minute and re-run this cell.")

def say(prompt, **kw):
    """Same as ask(), but returns just the text of the answer."""
    return (ask(prompt, **kw).choices[0].message.content or "").strip()

print("✅ Setup complete. Model:", MODEL)
````

````python
# ✏️ YOUR TURN 9a — change the question, then run
MY_QUESTION = "In two sentences, why do businesses care about customer churn?"

response = ask(MY_QUESTION, max_tokens=150)
print(response.choices[0].message.content)
print()
print(f"Tokens used: {response.usage.prompt_tokens} in, {response.usage.completion_tokens} out")
````

````python
# ✏️ YOUR TURN 9b — change the one-line instruction (the "system prompt"), run it, and compare
SYSTEM = "You are a blunt CFO. Answer in one sentence and always mention cost."

response = ask(MY_QUESTION, system=SYSTEM, max_tokens=150)
print(response.choices[0].message.content)
````
