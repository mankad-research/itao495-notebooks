## Cell 3 — Get your free Groq API key, and store it safely

We use **Groq** for most labs this semester because it is free and needs no credit card.

**Part A — create the key (2 minutes)**
1. Go to **https://console.groq.com** and sign up (Google sign-in works).
2. In the left sidebar, click **API Keys** → **Create API Key**.
3. Copy the key. It starts with `gsk_`.

### One-time setup: store your key in Colab Secrets
1. Click the **🔑 key icon** in the left sidebar of Colab.
2. Click **Add new secret**. Name: `GROQ_API_KEY`. Value: your key (it starts with `gsk_`).
3. Turn on **Notebook access** for this notebook.

Colab will remember the key for all your notebooks, so you never paste it into a cell. If you skip this, the setup cell will simply ask you to paste the key each time (the box hides what you type).

⚠️ **Treat the key like a password.** Do not paste it into a cell, post it publicly, or share a notebook that has it written inside. If you ever expose one, delete it in the Groq console and make a new one; it takes 10 seconds.

```python
from groq import Groq

def get_key():
    """Look for the key in Colab Secrets first; otherwise ask for it (hidden as you type)."""
    try:
        from google.colab import userdata
        return userdata.get("GROQ_API_KEY")
    except Exception:
        import getpass
        return getpass.getpass("Paste your Groq API key (starts with gsk_): ")

GROQ_API_KEY = get_key()

if GROQ_API_KEY and GROQ_API_KEY.startswith("gsk_"):
    client = Groq(api_key=GROQ_API_KEY)
    print("✅ Cell 3 passed — key found and client created.")
else:
    print("❌ No valid key yet. Follow Part A and the Colab Secrets steps above, then re-run this cell.")
```

## Cell 4 — Your first model response

This is a real call to a real large language model. It should come back in about a second.

```python
# Groq's free model. It replaced llama-3.1-8b-instant, which Groq retired in August 2026.
MODEL = "openai/gpt-oss-20b"

response = client.chat.completions.create(
    model=MODEL,
    messages=[{"role": "user", "content": "In one sentence, what is a business school?"}],
    max_tokens=400,            # room for a short hidden "thinking" step plus the answer
    reasoning_effort="low",    # think briefly: faster and cheaper
    include_reasoning=False,   # show only the final answer
)

print(response.choices[0].message.content)
print()
print("✅ Cell 4 passed — you just called an LLM from code. You are ready for Session 2.")
```

## 🛠️ Troubleshooting

| What you see | What it means | Fix |
|---|---|---|
| `AuthenticationError` / `invalid_api_key` | Key is wrong or has a typo | Re-copy it from console.groq.com; make sure no spaces came along |
| Cell 3 prints ❌ | No key found | Check the secret is named exactly `GROQ_API_KEY` and **Notebook access** is switched on |
| `NameError: name 'client' is not defined` | Cell 3 did not run successfully | Re-run Cell 3 and look for the ✅ |
| `ModuleNotFoundError: groq` | Cell 2 did not run, or the session restarted | Re-run Cell 2 |
| `RateLimitError` | Too many requests too fast (free tier) | Wait about 30 seconds and retry |
| `model_decommissioned` or `model not found` | Groq retired the model name | Check console.groq.com/docs/models and change `MODEL` in Cell 4 |
| Everything looks frozen | Colab disconnected | Runtime → Restart session, then re-run from Cell 1 |
